## Columns of Tables are Arrays ##

```python
nba = Table.read_table('nba_salaries.csv').relabeled(4, 'SALARY')
warriors = nba.where('team', 'Golden State Warriors')
```

```python
warriors
```

```python
salary_select = warriors.select('SALARY')
salary_select
```

```python
type(salary_select)
# Returns a column in a table format
```

```python
salary_col = warriors.column('SALARY')
salary_col
```

```python
type(salary_col)
```

```python
#np.average(salary_select)
# can't call numpy function on tables
```

```python
np.average(salary_col)
```

## Practice

Find the total salary paid to Chicago Bulls players in the year 2010.
- Use where to narrow the table to the right team and year.
- Use column to extract the correct column as an array.
- Use the sum command on the array.

```python
bulls = nba.where('team', 'Chicago Bulls')
bulls2010 = bulls.where('season', 2010)
#bulls2010.show()
salary_arr = bulls2010.column('SALARY')
np.average(salary_arr)

```
