## Defining Functions ##  

Example: Create a function that takes a numerical input and triples it: $\textsf{triple}(x)=3\,x$

````python
# square is the function name
# x is the input
# result is the output
def square(x):
    result = x**2
    return result
````

````python
def triple(x):
    result = 3 * x
    return result
````

````python
output = triple(4)
output
````

````python
num = 2
square(num)
````

### Practice

Example: Calculate the Hypotenuse Length of a Right Triangle


Pythagoras's Theorem: If $x$ and $y$ denote the lengths of the right-angle sides, then the hypotenuse length $h$ satisfies:

$$ h^2 = x^2 + y^2 \qquad \text{which implies}\qquad \hspace{20 pt} h = \sqrt{ x^2 + y^2 } $$

Write a function called hypotenuse that takes the lengths of the two legs of a right triangle and returns the length of the hypotenuse. You can use:

    ** 0.5

to calculate the square root of a number.

````python
def hypotenuse(leg1, leg2):
    sum_of_squares = leg1 ** 2 + leg2 ** 2
    hyp = sum_of_squares ** 0.5
    return hyp
````

````python
h = hypotenuse(3,4)
h
````

## Apply ##

````python
ages = Table().with_columns(
    'Person', make_array('Jim', 'Pam', 'Michael', 'Creed'),
    'Birth Year', make_array(1985, 1988, 1967, 1904)
)
ages
````

````python
def name_and_birthyear(name, year):
    out_string = name + " was born in " + str(year) + "."
    return out_string
````

````python
name_and_birthyear('Jim', 1985)
````

````python
# .apply allows you to specify a function and the *names* of columns in a table
# The specified function will be applied to each row of the table using inputs from the columns you specified
ages_and_years = ages.apply(name_and_birthyear, 'Person', 'Birth Year')
ages_and_years
````

````python
ages_and_years.item(0)
````

````python
ages = ages.with_column('Ages and Years', ages_and_years)
ages
````

# Practice

1. Define a function named percent_increase that takes two inputs: small and big. The function returns the the *percent* increase from small to big. For example, if small = 200 and big = 210, then percent_increase(small, big) returns 5.

2. Apply this function to the Gross and Gross (Adjusted) columns of top_movies.

3. Attach this new column to the table.

````python
top_movies = Table.read_table('top_movies_2017.csv').select("Title", "Gross", "Gross (Adjusted)")
top_movies
````

````python
def percent_increase(small, big):
    increase = round((big / small - 1) * 100)
    return increase

````

````python
percent_increase(198676459, 1796176700)
````

````python
growth_arr = top_movies.apply(percent_increase, 'Gross', 'Gross (Adjusted)')
growth_arr
````

````python
top_movies = top_movies.with_column('Percent Growth', growth_arr)
top_movies.show(10)
````
