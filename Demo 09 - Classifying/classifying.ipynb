# Classifying

```python
from datascience import *
import numpy as np

%matplotlib inline
import matplotlib.pyplot as plots
plots.style.use('fivethirtyeight')
```

```python
cones = Table().with_columns(
    'Flavor', make_array('strawberry', 'chocolate', 'chocolate', 'strawberry', 'chocolate'),
    'Price', make_array(3.55, 4.75, 6.55, 5.25, 5.25)
)
cones
```

```python
# Default behavior is to count the number of 
# occurrences of each category
cones.group('Flavor')
```

```python
# Optionally, you can specify any function you want to see a different 
# characteristic of the categories.
# This computes the most expensive ice cream in each category
cones.group('Flavor', max)
```

```python
# This computes the average of all prices in the category
cones.group('Flavor', np.average)
```

```python
nba = Table.read_table('nba_salaries.csv')
nba
```

# Practice 1
Find the total amount of money each team paid out in salaries in the year 2000.
- Select only the rows pertaining to the year 2000.
- Select only the columns relevant to your calculation.
- Group in such a way that computes the requested.

```python
nba_2000 = nba.where('season', 2000)
nba_2000 = nba_2000.select('team', 'salary')
salary_by_team = nba_2000.group('team', sum)
salary_by_team.show()
```

# Practice 2
Find the average salary earned by players in each position in the year 2000. (For example, the row with position 'C' contains the average salary paid to all centers across all teams in the year 2000.)

```python
nba_2000 = nba.where('season', 2000)
positions = nba_2000.select('position', 'salary')
average_by_position = positions.group('position', np.average)
average_by_position.show()
```

# Classifying by multiple variables

```python
more_cones = Table().with_columns(
    'Flavor', make_array('strawberry', 'chocolate', 'chocolate', 'strawberry', 'chocolate', 'bubblegum'),
    'Color', make_array('pink', 'light brown', 'dark brown', 'pink', 'dark brown', 'pink'),
    'Price', make_array(3.55, 4.75, 5.25, 5.25, 5.25, 4.75)
)

more_cones
```

```python
# Can provide a list of multiple categories to group by
more_cones.group(['Flavor', 'Color'])
```

```python
# Just like with a single variable, you can provide 
# a function to look at a different characteristic
more_cones.group(['Flavor', 'Color'], max)
```

# Practice 3
Find the total paid in salaries by each team *during each season*. 

- Select the columns relevant to your calculation.
- Group by multiple columns in such a way that the sum is returned.

```python
nba_salaries = nba.select('team', 'season', 'salary')
nba_salaries = nba_salaries.group(['team', 'season'], sum)
nba_salaries
```
